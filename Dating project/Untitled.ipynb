```python
corr = df.corr(numeric_only=True)
corr
```

```python
df.head()
```

```python
df.drop(columns=["male_decision","female_decision","age_gap"], inplace=True)
```

```python
numeric_cols = df.select_dtypes(include="number").columns
numeric_cols
```
