# Data Cleaning & Preprocessing

```python
data_cleaned= df.copy()
data_cleaned
```

```python
data_cleaned.dtypes
```

```python
data_cleaned["sex"].value_counts()
data_cleaned["smoker"].value_counts()
```

```python
data_cleaned.drop_duplicates(inplace=True)
```

```python
data_cleaned.shape
```

```python
data_cleaned["sex"] = data_cleaned["sex"].map({"male":0, "female":1})
```

```python
data_cleaned
```

```python
data_cleaned["smoker"] = data_cleaned["smoker"].map({"no":0, "yes":1})
data_cleaned
```

```python
data_cleaned["smoker"].value_counts()
```

```python
data_cleaned.rename(columns={
    "sex":"is_female",
    "smoker":"is_smoker" }, inplace=True)
```

```python
data_cleaned = pd.get_dummies(data_cleaned, columns=["region"], dtype=int, drop_first=True)
```

# Feature Engineering and Extraction

```python
data_cleaned["bmi_category"] = pd.cut( data_cleaned["bmi"], bins=[0,18.5,24.9,29.9,float('inf')], labels=['Underweight', 'normal', 'overweight', 'obease']
                                     )
data_cleaned
```

```python
data_cleaned = pd.get_dummies(data_cleaned, columns=["bmi_category"], prefix=None, dtype=int)
data_cleaned
```

```python
columns = ['age','bmi', 'children']
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data_cleaned[columns] = scaler.fit_transform(data_cleaned[columns])
data_cleaned
```
